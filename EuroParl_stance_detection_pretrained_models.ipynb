```python
from transformers import pipeline

classifier = pipeline("zero-shot-classification", model=model, tokenizer=tokenizer)

hypothesis_template = "This example is {}"
classes_verbalized = ["procedural", "factual and neutral", "a stance"]

df['predicted_label'] = df['en'].apply(lambda x: classifier(x, classes_verbalized, hypothesis_template=hypothesis_template, multi_label=False))
```

```python
def extract_stance_score(row):
    if 'a stance' in row['labels']:
        idx = row['labels'].index('a stance')  # Find the index of 'a stance'
        return row['scores'][idx]  # Return the score at that index
    return None  # Return None if 'a stance' is not found

# Apply the function to the 'info' column
df['a_stance_score'] = df['predicted_label'].apply(extract_stance_score)
```
