```python
df = pd.read_csv("data/train-00000-of-00001.csv")
df["EU Party"] = df["EU Party"].apply(lambda x: x.split('/')[-1])
```

```python
df["EU Party"].value_counts()
```

```python
df_party = df[df["en"].str.contains(r"\bparty\b", case=False, na=False)]
df_party["en"].to_csv(
    "speeches_containing_party.csv",
    sep=";",
    index=False,
    encoding="utf-8-sig"
)
```

```python
df = df[~df["EU Party"].str.contains(r"\bNA\b", case=False, na=False)]
majority = df["EU Party"].value_counts(normalize=True)
print("Majority baseline:", majority)
```
